```python
import pandas as pd

# Download public CSV of Amazon Electronics reviews
url = (
  "https://raw.githubusercontent.com/selva86/datasets/master/amazon_baby_subset.csv"
)

# This has columns: review, rating (1–5)
df = pd.read_csv(url)

# Sample 10K rows for your notebooks
df = df.sample(n=10000, random_state=42).reset_index(drop=True)

# Create binary sentiment label: 4–5 positive, else negative
df["label"] = (df["rating"] >= 4).astype(int)

print(df[["review", "rating", "label"]].head())
```

```python
from datasets import load_dataset

# Load 20k sample reviews from Electronics category
ds = load_dataset(
    "McAuley-Lab/Amazon-Reviews-2023",
    "raw_review_Electronics",
    split="full",
    trust_remote_code=True
)

# Sample for lightweight usage
ds_small = ds.shuffle(seed=42).select(range(20000))
df = ds_small.to_pandas()
print(df.head()[["text", "rating"]])
```

```python
import tensorflow_datasets as tfds
import pandas as pd

ds = tfds.load('imdb_reviews', split='train', as_supervised=True)

texts, labels = [], []
for text, label in ds: # .take(10000)
    texts.append(text.numpy().decode("utf-8"))
    labels.append(int(label.numpy()))

df = pd.DataFrame({"text": texts, "sentiment": labels})
df.to_csv("imdb.csv", index=False)
```
