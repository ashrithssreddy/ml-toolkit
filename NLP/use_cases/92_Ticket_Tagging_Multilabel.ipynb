#### 🏷️ Problem Statement: Multi-label Ticket Classification


You’re working on a customer support automation tool. Every day, a company receives hundreds of customer service tickets, which agents manually tag with issue types like:  
**"billing", "login issue", "technical bug", "refund", "delivery", "product feedback", etc.**

Manual tagging is slow and error-prone.

Your task is to **train a multi-label classifier** that takes in raw ticket text and **predicts multiple relevant tags** for each ticket, helping route it to the right department instantly.

This will:
- Reduce response time
- Improve agent efficiency
- Allow automated analytics on ticket trends
