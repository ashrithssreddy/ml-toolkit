# 🔍 Problem Statement
FAQ Matching with Semantic Search

You're building a smart assistant for a fintech app. Customers often ask questions via chat like:

- “How do I reset my password?”
- “Why was I charged twice?”
- “What is your refund policy?”

Instead of manually responding, you want to **automatically match** each query to the most relevant answer from an internal FAQ bank using **embedding-based semantic similarity**.

Your job is to:
- Embed both queries and FAQs
- Find the top semantic match
- Enable efficient vector search in production
