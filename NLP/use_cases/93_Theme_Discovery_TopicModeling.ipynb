# 🧠 Problem Statement
Discovering Themes in Amazon Reviews

You're helping a product manager at Amazon make sense of thousands of customer reviews for a new smart speaker.

Rather than reading each review manually, they want to **automatically extract key themes** (topics) from the reviews to understand:

- What are customers praising?
- What are the common complaints?
- Are there emerging concerns post-launch?

Your task is to apply **topic modeling** (LDA/NMF) to uncover hidden structure and provide an interpretable summary of customer feedback.
